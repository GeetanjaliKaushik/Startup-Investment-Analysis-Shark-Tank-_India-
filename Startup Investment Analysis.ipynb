```python
# 3. Check basic structure
print("Shape of dataset:", df.shape)
print("\nColumn Names:\n", df.columns.tolist())
print("\nMissing Values:\n", df.isnull().sum())
```

```python
# 4. Convert important columns to numeric
numeric_cols = [
    'Total Deal Amount', 'Total Deal Equity', 'Total Deal Debt',
    'Original Ask Amount', 'Original Offered Equity', 'Valuation Requested',
    'Deal Valuation', 'Yearly Revenue', 'Monthly Sales',
    'Gross Margin', 'Net Margin', 'Number of Presenters',
    'Male Presenters', 'Female Presenters', 'Transgender Presenters',
    'Couple Presenters', 'Pitchers Average Age', 'Number of sharks in deal'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
# 5. Create funding success column
df['Funding Success'] = df['Total Deal Amount'].apply(lambda x: 1 if pd.notnull(x) and x > 0 else 0)
```

```python
# check which startups got funding, use
funded_startups = df[df['Total Deal Amount'] > 0]
print(funded_startups[['Startup Name', 'Industry', 'Total Deal Amount']].head())
```

```python
# 6. Clean offer columns if they are Yes/No type
df['Received Offer'] = df['Received Offer'].astype(str).str.strip().str.lower()
df['Accepted Offer'] = df['Accepted Offer'].astype(str).str.strip().str.lower()
```

```python
# 7. Basic project KPIs
total_startups = df['Startup Name'].nunique(dropna=False)
total_pitches = df['Pitch Number'].nunique(dropna=False)
total_funding = df['Total Deal Amount'].sum()
funded_startups = df[df['Funding Success'] == 1]['Startup Name'].nunique(dropna=False)
success_rate = (funded_startups / total_startups) * 100 if total_startups > 0 else 0
avg_deal_amount = df['Total Deal Amount'].mean()

print("\n------ KPI SUMMARY ------")
print("Total Startups:", total_startups)
print("Total Pitches:", total_pitches)
print("Total Funding:", total_funding)
print("Funded Startups:", funded_startups)
print("Funding Success Rate:", round(success_rate, 2), "%")
print("Average Deal Amount:", round(avg_deal_amount, 2))
```

```python
# 8. Industry-wise investment analysis
industry_funding = df.groupby('Industry', as_index=False)['Total Deal Amount'].sum().sort_values(by='Total Deal Amount', ascending=False)
print("\n------ Industry-wise Total Funding ------")
print(industry_funding.head(10))
```

```python
# 9. Industry-wise deal count
industry_deals = df[df['Funding Success'] == 1].groupby('Industry', as_index=False)['Startup Name'].count()
industry_deals.rename(columns={'Startup Name': 'Funded Startups Count'}, inplace=True)
industry_deals = industry_deals.sort_values(by='Funded Startups Count', ascending=False)
print("\n------ Industry-wise Funded Startup Count ------")
print(industry_deals.head(10))
```

```python
 # 10. Founder success pattern
founder_success = df.groupby('Number of Presenters', as_index=False)['Funding Success'].mean()
founder_success['Funding Success %'] = founder_success['Funding Success'] * 100
print("\n------ Founder Count vs Success Rate ------")
print(founder_success)
```

```python
# 11. Gender-based presenter analysis
gender_analysis = df.groupby(['Male Presenters', 'Female Presenters'], as_index=False)['Total Deal Amount'].sum()
print("\n------ Gender-based Funding Pattern ------")
print(gender_analysis.head(10))
```

```python
# 12. State-wise startup count
state_analysis = df.groupby('Pitchers State', as_index=False)['Startup Name'].count()
state_analysis.rename(columns={'Startup Name': 'Startup Count'}, inplace=True)
state_analysis = state_analysis.sort_values(by='Startup Count', ascending=False)
print("\n------ State-wise Startup Count ------")
print(state_analysis.head(10))
```

```python
# 13. Revenue vs funding
revenue_funding = df[['Startup Name', 'Industry', 'Yearly Revenue', 'Monthly Sales', 'Total Deal Amount']].sort_values(by='Total Deal Amount', ascending=False)
print("\n-- Revenue vs Funding --")
print(revenue_funding.head(10))
```

```python
# 14. Investor analysis
investor_cols = [
    'Ashneer Investment Amount',
    'Namita Investment Amount',
    'Anupam Investment Amount',
    'Vineeta Investment Amount',
    'Aman Investment Amount',
    'Peyush Investment Amount',
    'Ghazal Investment Amount',
    'Amit Investment Amount',
    'Guest Investment Amount'
]

for col in investor_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

investor_totals = pd.DataFrame({
    'Investor': investor_cols,
    'Total Investment': [df[col].sum() for col in investor_cols]
})

investor_totals['Investor'] = investor_totals['Investor'].str.replace(' Investment Amount', '', regex=False)
investor_totals = investor_totals.sort_values(by='Total Investment', ascending=False)

print("\n------ Investor-wise Total Investment ------")
print(investor_totals)
```

```python
# 15. Deals with conditions
condition_analysis = df.groupby('Deal has conditions', as_index=False)['Startup Name'].count()
condition_analysis.rename(columns={'Startup Name': 'Count'}, inplace=True)
print("\n------ Deals with Conditions ------")
print(condition_analysis)
```

```python
# 16. Patent analysis
patent_analysis = df.groupby('Has Patents', as_index=False)['Funding Success'].mean()
patent_analysis['Funding Success %'] = patent_analysis['Funding Success'] * 100
print("\n------ Patent vs Funding Success ------")
print(patent_analysis)
```

```python
# 17. Top funded startups
top_startups = df[['Startup Name', 'Industry', 'Total Deal Amount', 'Deal Valuation']].sort_values(by='Total Deal Amount', ascending=False)
print("\n------ Top Funded Startups ------")
print(top_startups.head(10))
```
